```python
# استيراد مكتبة PyTorch
import torch

# استيراد مكتبة التعامل مع الصور
from PIL import Image

# استيراد مكتبة عرض الصور
import matplotlib.pyplot as plt

# استيراد مكتبة التعامل مع البيانات
import numpy as np

print("تم تجهيز المكتبات بنجاح!")
```

```python
from transformers import pipeline

# تحميل نموذج تصنيف الصور
model = pipeline(
    "image-classification",
    model="google/vit-base-patch16-224"
)

print("تم تحميل نموذج تصنيف الصور بنجاح!")
```

```python
from google.colab import files

# رفع صورة من الجهاز
uploaded = files.upload()

# الحصول على اسم الصورة
image_path = next(iter(uploaded))

# فتح الصورة
image = Image.open(image_path).convert("RGB")

# عرض الصورة
plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
# تحليل الصورة باستخدام نموذج الذكاء الاصطناعي
results = model(image)

# عرض أفضل 3 توقعات
for result in results[:3]:
    label = result["label"]
    score = result["score"] * 100

    print("التصنيف:", label)
    print("نسبة الثقة:", round(score, 2), "%")
    print("-" * 30)
```
